```python
import numpy as np


x = np.array([
    [1,2,3],
    [2,3,4],
    [3,4,5],
    [4,5,6],
])

w = np.array([10,20,30])
```

```python
# We coudl also do a copy.deepcopy
y = np.zeros(x.shape)


# x.shapre returns the number of rows anc olumns in a tuple
# So x.shape[0] is the number of rows
for i in range(x.shape[0]): 
    y[i,:] = x[i,:] + w

y
```

```python
# Broadcasting

# Row additon, linear algebra operation. They dont need to be the same size
z = x + w

print(z)
```

```python
z1 = x * w

z1
```

```python
# Column vector

# Transpose a row into a column
v = np.array([-1,0,1,0], ndmin=2).T

v
```

```python
print(x)
print(" ")
x*v
```

```python
# Excercise
# Create a vector of integers

v = np.array(range(0,9))
print(v)
# Reshape v vector into a 3x3 matrix (M) (np.reshape)

m = np.reshape(v,(3,3))
print(m)

# Repeat m into a 9x3 c

c = np.tile(m,(3,1))
print(c)

# Broadcast

b = c*np.reshape(v,(len(v),1))
b
```
