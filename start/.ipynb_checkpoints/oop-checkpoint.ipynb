```python
# A class is a blueprint for creating a set of atributes and methods
```

```python
class model(object):
    # Constrtor method
    def __init__(self, numlayers,numunits,name):
        # Set attributes
        self.layers = numlayers
        self.units = numunits
        self.name = name
        self.weights = 1

    def howManyUnits(self):
        totalUnits = self.layers * self.units
        print(f"There are {totalUnits} in the model")

    def trainModel(self,x):
        self.weights += x
        return self.weights

    def __str__(self):
        return f'This is a {self.name} architecture'
```

```python
m1 = model(12,5,'CNN')

m1.howManyUnits()
```

```python
m1.trainModel(10)
```

```python
m1.__str__()
```

```python
import numpy as np

class model2(object):
    # Constrtor method
    def __init__(self, numlayers,numunits,name):
        # Set attributes
        self.layers = numlayers
        self.units = numunits
        self.name = name
        self.weights = np.random.randn(numlayers,numunits)

    def howManyUnits(self):
        totalUnits = self.layers * self.units
        print(f"There are {totalUnits} in the model")

    def trainModel(self,x,y):
        self.weights = self.weights * x + y        

    def __str__(self):
        return f'This is a {self.name} architecture'
```

```python
m2 = model2(2,3,'x')

m2.weights
```

```python
m2.trainModel(2,3)

m2.weights
```
