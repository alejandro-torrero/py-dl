```python
def sumParams (x,y: int) -> int:
    return x + y
```

```python
sumParams(2,2)
```

```python
def sumAndProduct(x,y:int):
    return x+7,x*y
```

```python
sumAndProduct(2,3)
```

```python
sumR,prodR = sumAndProduct(10,3)
print(sumR,prodR)
```

```python
# Factorial function

def factorial(x):
    if x == 1:
        return 1
    return x * factorial(x-1)

factorial(9)
```

```python
import numpy as np

# Coin function

def flipCoin(n:int):
    flip_results = np.random.randn(n) >0 
    return np.mean(flip_results)

flipCoin(1000000)
```
