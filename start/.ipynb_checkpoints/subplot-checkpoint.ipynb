```python
import matplotlib.pyplot as plt
import numpy as np

# rows, columns index of the plot
# Divides the workspace into subplots
plt.subplot(2,2,1)
plt.plot(np.random.randn(10))

plt.subplot(2,2,4)
plt.plot(np.random.randn(10))

plt.show()
```

```python
fig,ax = plt.subplots(1,2,figsize=(15,3))

x = np.arange(10)

ax[0].plot(x,x**2,'b')
ax[0].plot(x,x)
ax[1].plot(x,np.sqrt(x),'r')
print(ax.shape)
plt.show()
```

```python
fix,ax = plt.subplots(2,2)

ax[0,0].plot(np.random.randn(4,4))
ax[0,1].plot(np.random.randn(4,4))
ax[1,0].plot(np.random.randn(4,4))
ax[1,1].plot(np.random.randn(4,4))

plt.tight_layout()

plt.show()
```

```python
# 3 by 3 subplot geomtry in a for loop

fix,ax = plt.subplots(3,3)

for i in range(3):
    for j in range(3):
        ax[i,j].plot(np.random.randn(2,2))

plt.tight_layout()
plt.show()
```
