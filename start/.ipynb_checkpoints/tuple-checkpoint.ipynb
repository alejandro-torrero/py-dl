```python
# Tuple

# A tuple is immutable. You cannot change the values after you define it
atuple = (3, "a", 1.0)

# A list is mutable
alist = [3, "a", 1.0]

print(atuple, alist)
```

```python
# Edit a list

print(alist)
alist[0] = 13.13
print(alist)
```

```python
x = 10
print(id(x))

x = 20
print(id(x))
```
