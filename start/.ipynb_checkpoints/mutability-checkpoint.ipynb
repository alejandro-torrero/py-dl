```python
a = 10
b = a # Creates a new reference as this kind of variables is immutable

print(a,b)

b = 5

print(a,b)
```

```python
arr_a = [1,2,3]

arr_b = arr_a
print(arr_a, arr_b)

arr_b[0] = 500

print(arr_a, arr_b)
```

```python
import copy 

obj = {"name":"Pedro", 'age':2}

obj_copy = obj
print(id(obj))
print(id(obj_copy))

obj_deep = copy.deepcopy(obj)

print(id(obj))
print(id(obj_deep))
```
