```python
import numpy  as np
import pandas as pd

# Pandas is a library to handle datasets that can be organized as observations by features
# Data that you would put into a 2 dimensional schema, also higher dimensions can be used

# Create random data

var1 = np.random.randn(100) *5 +20
var2 = np.random.randn(100) > 0


# Variable labels

labels = ["Temp (C)", "Ice cream"]

# Put data together into a dictionary

D = {labels[0]: var1, labels[1]: var2}


# Import the dictionary into a pandas data frame

df = pd.DataFrame(data=D)
```
