```python
# Simple function

def prod(x,y: int):
    return x * y

prod(2,3)
```

```python
# Overloaded operation

prod(2,"hi")
```

```python
# Ad type check

def prodV(x,y):
    if not isinstance(x,(int,float)):        
        if isinstance(x,str) and x.isnumeric():
            x = float(x)
            print('Converted x')
        else:
            raise Exception("Input x must be a number")

    if not isinstance(y,(int,float)):
        raise Exception("Input y must be an integer")

    return x * y
        
```

```python
prodV(10,'hi')
```

```python
prodV('2',3)
```
