```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Create a list

X  = [1,2,4,6,5,4,0,-1,5,-2,6,10,-9,1,3,-6]
n = len(X)

# Compute population mean
popmean = np.mean(X)

# Get a random sample fo the list
sample = np.random.choice(X,size=5,replace=True)
sampmean= np.mean(sample)

print(popmean)
print(sampmean)
```

```python
# Compute test with a larger data set

m = 10000

sampleMeans = np.zeros(m)

# Build m samples
for i in range(m):
    sample = np.random.choice(X,size=15,replace=True)
    sampleMeans[i] = np.mean(sample)

plt.hist(sampleMeans,bins=40, density=True)
plt.plot([popmean,popmean],[0,0.3],'m--')
plt.ylabel("Count")
plt.xlabel("Sample mean")
plt.show()
```
