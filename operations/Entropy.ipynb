```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Probability of an event happening

p = .25

# Not correct example

H = - (p*np.log(p))
print(H)

# we need to also cosnidering the probaility of the event p not happening, represented by q
q = 1 - p

H = - ((p*np.log(p))+(q*np.log(q)))
print(H)
```

```python
# Possible events 
X = [0.25,0.75]

H = 0

for p in X:
    H -= p*np.log(p)

# Entropy of an event with a probabilty of 0.25
print("Entropy: ", H)
```

```python
# Going back to the two events

# The probabilty of an event happening is 0.3
p = 0.3

# Then the cross entropy would be
# This form represents the cross entropy of the two events: happening and not happening
# Evaluating only two events
H = - ((p*np.log(p))+((1-p)*np.log((1-p))))
print(H)
```

```python
# Cross entropy example

# Lets say we have a model that detects cats, then we have two scenarios, detecting a cat or not

# Represented in p, we can have a 1 for a cat detected, or 0 otherwise
p = [1,0] # 

# Then the probabilty of each event

q = [0.25,0.75]

H = 0
for i in range(len(p)):
    H -= p[i]*np.log(q[i])

print("Cross entropt",H)
```

```python
# Using pytorch

import torch
import torch.nn.functional as F

q_tensor = torch.Tensor(q)
p_tensor = torch.Tensor(p)

F.binary_cross_entropy(q_tensor,p_tensor)
```
