```python
import numpy as np
import matplotlib.pyplot as plt

import sympy as sym
import sympy.plotting.plot as symplot
```

```python
# Create synbolic variables in sympy

x = sym.symbols('x')

# Create a function

fx = 2*x**2

# Compute derivative

df = sym.diff(fx,x)

print(fx)
print(df)
```

```python
# Plot function

symplot(fx,(x,-4,4),title='The function')
plt.show()

symplot(df,(x,-4,4), title="Its derivative")
plt.show()
```
