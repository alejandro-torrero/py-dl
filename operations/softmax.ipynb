```python
import numpy as np
import torch
import torch.nn as nn
import matplotlib.pyplot as plt
```

```python
# Manually in numpy

z = [1,2,3]

# Compute softmax result

num =  np.exp(z)
den = np.sum(np.exp(z))

sigma = num/den

print(sigma)
```

```python
# Repeat with random integers

z = np.random.randint(-5,high=15,size=25)

num = np.exp(z)
den = np.sum(np.exp(z))

sigma = num/den

print(sigma)
print()
sum(sigma)

# Plot

plt.plot(z,sigma,'ko')
plt.xlabel("Original number (z)")
# Latec code
print("$\sum\sigma$")
plt.ylabel("Softmaxified $\sigma$")
plt.title("$\sum\sigma$ = %g" %np.sum(sigma))
plt.show()
```

```python
# Pytorch 

z = [1,2,3]

softClass = nn.Softmax(dim=0)

sigmaT = softClass(torch.tensor(z,dtype=torch.float32))

print(sigmaT)
```

```python
# Plot pytorch

plt.plot(z,sigmaT,"ko")
plt.show()
```
