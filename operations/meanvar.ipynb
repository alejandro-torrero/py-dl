```python
import numpy as np
```

```python
X = [1,2,4,6,5,4,0]

n = len(X)

mean1 = np.mean(X)
mean2 = np.sum(X) / n

print(mean1)
print(mean2)
```

```python
# Variance

var1 = np.var(X) # By default it has ddof=0
var2 = (1/(n-1)) * np.sum((X-mean1)**2)

print(var1)
print(var2)
```

```python
# Why do we get different values for variance?

# We  need to specifcy degrees of freedom

var3 = np.var(X,ddof=1)
print(var3)
```

```python
# Does ddof matter for larger sets?

m = 100000
z = np.random.randint(0, high = 20, size=m)

var0 = np.var(z,ddof=0) # default
var1 = np.var(z,ddof=1) # unbiased

print(var0)
print(var1)
```
