```python
import numpy as np
import torch
```

```python
# numpy 

# Create an array, or a vector on linear algebra
nv = np.array([[1,2,3,4]])
print(nv)
print(" ")

# Transpose vector
print(nv.T)
print(" ")

# Transpose again
nvT = nv.T

print(nvT.T)
print(" ")
```

```python
# A matrix

# Create a matrix
nM = np.array([[1,2,3,4],[5,6,7,8]])
print(nM)
print(" ")

# Transpose vector
print(nM.T)
print(" ")

# Transpose again
nmT = nM.T

print(nmT.T)
print(" ")
```

```python
# Using pytorch

tv = torch.tensor([[1,2,3,4]])

print(tv)
print("")

# Transpose
print(tv.T)
print("")

# Transpose again
tvT = tv.T
print(tvT.T)
print("")
```

```python
# torhc with matrix
tm = torch.tensor([[1,2,3,4],[5,6,7,8]])

print(tm)
print("")

# Transpose
print(tm.T)
print("")

# Transpose again
tmT = tm.T
print(tmT.T)
print("")
```
