```python
import numpy as np
import torch
import torch.nn as nn
```

```python
# Arg min using numpy

v = np.array([1,40,2,-3])

# Find and report the maximum and minimum values

minVal = np.min(v)
maxVal = np.max(v)

print(f"Min: {minVal} Max: {maxVal}")

# Get arg max and arg min
minidx = np.argmin(v)
maxidx = np.argmax(v)

print(f"ArgMin at: {minidx} ArgMax at: {maxidx}")

# Confirm argmin and argmax

print(f"ArgMin val: {v[minidx]} ArgMax val: {v[maxidx]}")
```

```python
# Evaluating matrix

M = np.array([[0,1,10],[20,8,5]])

# A matrix has multiple ways to get its max and min values, the whole matrix can be evaluated, or a column, or a row

print(M)
print()

minvals1 = np.min(M) # Min value of the entire matrix
minvals2 = np.min(M,axis=0) # Min value of each column (across rows)
minvals3 = np.min(M,axis=1) # Min value of each row (across columns)

print(minvals1)
print(minvals2)
print(minvals3)
print()

# Argmin

mindidx1 = np.argmin(M) # Min value of the entire matrix
mindidx2 = np.argmin(M,axis=0) # Min value of each column (across rows)
mindidx3 = np.argmin(M,axis=1) # Min value of each row (across columns)

print(mindidx1)
print(mindidx2)
print(mindidx3)
```

```python
# Using torch

v = torch.tensor([1,40,2,-3])

# Find and report the maximum and minimum values

minVal = torch.min(v)
maxVal = torch.max(v)

print(f"Min: {minVal} Max: {maxVal}")

# Get arg max and arg min
minidx = torch.argmin(v)
maxidx = torch.argmax(v)

print(f"ArgMin at: {minidx} ArgMax at: {maxidx}")

# Confirm argmin and argmax

print(f"ArgMin val: {v[minidx]} ArgMax val: {v[maxidx]}")
```

```python
# Torch and amtrix

M = torch.tensor([[0,1,10],[20,8,5]])

print(M)
print()

min1 = torch.min(M)
min2 = torch.min(M,axis=0)
min3 = torch.min(M,axis=1)

print(min1)
print()
print(min2)
print(min2.values)
print(min2.indices)

print()
print(min3)
print(min3.values)
print(min3.indices)
```
