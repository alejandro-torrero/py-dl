```python
import numpy as np
import torch
```

```python
np.random.randn(5)
```

```python
# Get random numbers from the same seed
np.random.seed(17)
print(np.random.randn(5))
print(np.random.randn(5))
```

```python
# Local scope for the seed
randseed1 = np.random.RandomState(17)
randseed2 = np.random.RandomState(20210530)

print(randseed1.randn(5))
print(randseed2.randn(5))
print(randseed1.randn(5))
print(randseed2.randn(5))
print(np.random.randn(5))
```

```python
# pytorch

torch.randn(5)
```

```python
torch.manual_seed(17)
print(torch.randn(5))
print(torch.randn(5))
```
