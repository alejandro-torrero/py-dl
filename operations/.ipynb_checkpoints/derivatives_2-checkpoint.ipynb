```python
import numpy as np
import sympy as sym

%config InlineBackend.figure_formats = ['svg']
```

```python
# Create symbolic variables in sympy

x = sym.symbols('x')
fx = 2*x**2
gx = 4*x**3-3*x**4

print("Functions")
display(fx)
display(gx)


# Compute individual derivatives
print("Derivatives")
df = sym.diff(fx)
dg = sym.diff(gx)

display(df)
display(dg)


# Apply product rule manually

manual = df*gx + fx*dg
thewrongway = df*dg
print("Manual result")
display(manual)

# Using sympy
print("Sympy result")
sympyway= sym.diff(fx*gx)
display(sympyway)
```
