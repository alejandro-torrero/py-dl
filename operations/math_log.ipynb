```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Set of points

x = np.linspace(.0001,1,200)

logx = np.log(x)

# Plot results

fig = plt.figure(figsize=(10,4))

plt.rcParams.update({'font.size':15})

plt.plot(x,logx,'ks-',markerfacecolor='w')

plt.xlabel('x')
plt.ylabel('log(x)')

plt.show()
```

```python
logx = np.log(x)
expx = np.exp(x)

# Plot

plt.plot(x,x,color=[0.8,0.8,0.8])
plt.plot(x,np.exp(logx),'o')
plt.plot(x,np.log(expx),'x')

plt.xlabel('x')
plt.ylabel('f(g(x))')
plt.legend(['Unity','exp(log(x))','log(exp(x))'])

plt.show()

# They represent the same line!!
```
