# Gradient descent 2D

```python
import numpy as np
import matplotlib.pyplot as plt
import sympy as sym
```

```python
# Peaks function

def peaks(x,y):
    # Expand a 2d mesh
    x,y = np.meshgrid(x,y)

    z = 3*(1-x)**2 * np.exp(-(x**2)-(y+1)**2) \
        - 10*(x/5 - x**3 - y **5) * np.exp(-x**2-y**2) \
        -1/3 * np.exp(-(x+1)**2 - y**2)
    return z
```

```python
# Create landscape

x = np.linspace(-3,3,201)
y = np.linspace(-3,3,201)

z = peaks(x,y)


plt.imshow(z,extent=[x[0],x[-1],y[0],y[-1]], vmin=-5,vmax=5,origin='lower')
plt.show()
```

```python
# Create derivative functions using sym

sx,sy = sym.symbols('sx,sy')

sz = 3*(1-sx)**2 * sym.exp(-(sx**2)-(sy+1)**2) \
        - 10*(sx/5 - sx**3 - sy **5) * sym.exp(-sx**2-sy**2) \
        -1/3 * sym.exp(-(sx+1)**2 - sy**2)

# Create functions from the sympy computed derivatives

df_x = sym.lambdify((sx,sy),sym.diff(sz,sx),'sympy')
df_y = sym.lambdify((sx,sy),sym.diff(sz,sy),'sympy')

df_x(1,1).evalf()
```

```python
# Gradient descent algoritm

localmin = np.random.rand(2)*4-2
startpoint = localmin[:]

# Learning parameters
learning_rate = .01
training_epochs = 100

trajectory = np.zeros((training_epochs,2))

for i in range(training_epochs):
    grad = np.array([
        df_x(localmin[0],localmin[1]).evalf(),
        df_y(localmin[0],localmin[1]).evalf()
    ])

    localmin = localmin - learning_rate * grad
    trajectory[i:] = localmin

print(localmin)
print(startpoint)

# plot results

plt.imshow(z,extent=[x[0],x[-1],y[0],y[-1]], vmin=-5,vmax=5,origin='lower')
plt.plot(startpoint[0],startpoint[1],'bs')
plt.plot(localmin[0],localmin[1],'ro')
plt.plot(trajectory[:,0],trajectory[:,1],'r')
plt.legend(['rnd start','localmin'])
plt.colorbar()
plt.show()
```
