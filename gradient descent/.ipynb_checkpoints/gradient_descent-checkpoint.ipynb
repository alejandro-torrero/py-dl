```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Gradient descent in 1D

def fx(x):
    return 3*x**2-3*x +4

# Derivative function

def deriv(x):
    return 6*x -3
```

```python
# Plot the function and its derivative

x = np.linspace(-2,2,2001)

# Create plot

plt.plot(x,fx(x))
plt.plot(x,deriv(x))

plt.xlim(x[[0,-1]])

plt.grid()

plt.xlabel("x")
plt.ylabel("f(x)")
plt.legend(['y','dy'])

plt.show()
```

```python
# learning algorithm

# Pick one random value out of the vector x
localmin = np.random.choice(x,1)
print(localmin)

# learning parametres
learning_rate = .01
training_epochs = 100

# Run training

for i in range(training_epochs):
    grad = deriv(localmin)
    localmin = localmin - learning_rate*grad

print(localmin)
```

```python
# Plot the learnign results

plt.plot(x,fx(x),x,deriv(x))

plt.plot(localmin, deriv(localmin),'ro')
plt.plot(localmin, fx(localmin),'ro')

plt.xlim(x[[0,-1]])

plt.grid()

plt.xlabel("x")
plt.ylabel("f(x)")
plt.legend(['y','dy'])

plt.title("Empirical local minimum")

plt.show()
```

```python
# Get gradient descent while saving the local minima

# Pick one random value out of the vector x
localmin = np.random.choice(x,1)
print(localmin)

# learning parametres
learning_rate = .01
training_epochs = 100

# Run training
modelparams = np.zeros((training_epochs,2))

for i in range(training_epochs):
    grad = deriv(localmin)
    localmin = localmin - learning_rate*grad
    modelparams[i,0] = localmin[0]
    modelparams[i,1] = grad[0]
```

```python
# plot the computed localmin

fig,ax= plt.subplots(1,2,figsize=(12,4))

for i in range(2):
    ax[i].plot(modelparams[:,i],'o-')
    ax[i].set_xlabel('iteration')
    ax[i].set_title(f'Final estimated minimum {localmin[0]:0.5f}')

ax[0].set_ylabel('Local minimum')
ax[1].set_ylabel('Derivative')

plt.show()
```

```python
# Pick one random value out of the vector x
localmin = np.random.choice(x,1)
print(localmin)

# learning parametres
learning_rate = .01
training_epochs = 1000
learning_treshhold = 0.001

modelparams = np.zeros((training_epochs,2))

for i in range(training_epochs):
    grad = deriv(localmin)
    if abs(grad) <= learning_treshhold:
        print("Breaking cycle at epoch:",i)
        break
    localmin = localmin - learning_rate*grad
    modelparams[i,0] = localmin[0]
    modelparams[i,1] = grad[0]

# plot the computed localmin

fig,ax= plt.subplots(1,2,figsize=(12,4))

for i in range(2):
    ax[i].plot(modelparams[:,i],'o-')
    ax[i].set_xlabel('iteration')
    ax[i].set_title(f'Final estimated minimum {localmin[0]:0.5f}')

ax[0].set_ylabel('Local minimum')
ax[1].set_ylabel('Derivative')

plt.show()
```
